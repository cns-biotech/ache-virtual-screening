```python
from rdkit import Chem
from rdkit.Chem import Descriptors, Draw, AllChem, DataStructs
from rdkit import RDLogger
RDLogger.DisableLog("rdApp.*")

clean["mol"] = clean["canonical_smiles"].apply(Chem.MolFromSmiles)
print("не распарсилось:", clean["mol"].isna().sum())
```

```python
def strip_salt(mol):
    if mol is None:
        return None
    frags = Chem.GetMolFrags(mol, asMols=True)
    return max(frags, key=lambda m: m.GetNumAtoms())

n_multi = clean["canonical_smiles"].str.contains(r"\.").sum()
print("записей с несколькими фрагментами:", n_multi)

clean["mol"] = clean["mol"].apply(strip_salt)
clean = clean[clean["mol"].notna()].copy()
print("осталось молекул:", len(clean))
```

```python
clean["MW"]   = clean["mol"].apply(Descriptors.MolWt)
clean["LogP"] = clean["mol"].apply(Descriptors.MolLogP)
clean["HBD"]  = clean["mol"].apply(Descriptors.NumHDonors)
clean["HBA"]  = clean["mol"].apply(Descriptors.NumHAcceptors)
clean["TPSA"] = clean["mol"].apply(Descriptors.TPSA)
clean[["MW", "LogP", "HBD", "HBA", "TPSA", "pIC50"]].describe()
```

```python
!pip -q install rdkit
from rdkit import Chem
from rdkit.Chem import Descriptors
from rdkit import RDLogger
RDLogger.DisableLog("rdApp.*")

clean = pd.read_csv(f"{WORK}/ache_clean.csv")
clean["mol"] = clean["canonical_smiles"].apply(Chem.MolFromSmiles)
print("не распарсилось:", clean["mol"].isna().sum())

def strip_salt(mol):
    if mol is None:
        return None
    frags = Chem.GetMolFrags(mol, asMols=True)
    return max(frags, key=lambda m: m.GetNumAtoms())

print("с несколькими фрагментами:", clean["canonical_smiles"].str.contains(r"\.").sum())
clean["mol"] = clean["mol"].apply(strip_salt)
clean = clean[clean["mol"].notna()].copy()

clean["MW"]   = clean["mol"].apply(Descriptors.MolWt)
clean["LogP"] = clean["mol"].apply(Descriptors.MolLogP)
clean["HBD"]  = clean["mol"].apply(Descriptors.NumHDonors)
clean["HBA"]  = clean["mol"].apply(Descriptors.NumHAcceptors)
clean["TPSA"] = clean["mol"].apply(Descriptors.TPSA)

clean["lipinski_ok"] = (clean["MW"] <= 500) & (clean["LogP"] <= 5) & (clean["HBD"] <= 5) & (clean["HBA"] <= 10)

clean["smiles_clean"] = clean["mol"].apply(Chem.MolToSmiles)
print("уникальных структур после солей:", clean["smiles_clean"].nunique(), "из", len(clean))

clean = clean.groupby("smiles_clean", as_index=False).agg({
    "molecule_chembl_id": "first", "canonical_smiles": "first",
    "standard_value": "median", "pIC50": "median",
    "MW": "first", "LogP": "first", "HBD": "first",
    "HBA": "first", "TPSA": "first", "lipinski_ok": "first",
})
clean["mol"] = clean["smiles_clean"].apply(Chem.MolFromSmiles)

clean.drop(columns=["mol"]).to_csv(f"{WORK}/ache_descriptors.csv", index=False)

print("итого молекул:", len(clean))
print("LogP > 5:", (clean["LogP"] > 5).sum())
print("MW > 500:", (clean["MW"] > 500).sum())
print("Липински прошли:", clean["lipinski_ok"].sum(), "доля:", round(clean["lipinski_ok"].mean(), 3))
```

```python
from chembl_webresource_client.new_client import new_client
from rdkit.Chem import Draw

rec = new_client.molecule.get("CHEMBL502")
ref_smiles = rec["molecule_structures"]["canonical_smiles"]
print(rec["pref_name"], ref_smiles)

ref_mol = Chem.MolFromSmiles(ref_smiles)
Draw.MolToImage(ref_mol, size=(350, 300))
```

```python
from rdkit.Chem import rdFingerprintGenerator, DataStructs

gen = rdFingerprintGenerator.GetMorganGenerator(radius=2, fpSize=2048)

clean["fp"] = clean["mol"].apply(gen.GetFingerprint)
ref_fp = gen.GetFingerprint(ref_mol)

clean["tanimoto"] = clean["fp"].apply(lambda fp: DataStructs.TanimotoSimilarity(ref_fp, fp))
print(clean["tanimoto"].describe().round(3))
```

```python
top20 = clean.nlargest(20, "tanimoto")
print(top20[["molecule_chembl_id", "tanimoto", "pIC50", "MW", "LogP"]].round(3).to_string(index=False))

Draw.MolsToGridImage(
    list(top20["mol"]),
    molsPerRow=4,
    subImgSize=(260, 220),
    legends=[f"{r.molecule_chembl_id}  T={r.tanimoto:.2f}  pIC50={r.pIC50:.1f}"
             for r in top20.itertuples()],
)
```

```python
similar = clean[clean["tanimoto"] >= 0.4]
rest    = clean[clean["tanimoto"] <  0.4]

print("похожих (T >= 0.4):", len(similar), "медиана pIC50:", round(similar["pIC50"].median(), 2))
print("остальных:", len(rest), "медиана pIC50:", round(rest["pIC50"].median(), 2))
print("pIC50 донепезила в нашей выборке:",
      clean.loc[clean["molecule_chembl_id"] == "CHEMBL502", "pIC50"].values)
```

```python
ref_can = Chem.MolToSmiles(ref_mol)
hit = clean[clean["smiles_clean"] == ref_can]
print(len(hit))
print(hit[["molecule_chembl_id", "pIC50", "tanimoto"]])
print("максимум Танимото в выборке:", clean["tanimoto"].max())
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 2, figsize=(12, 4.5))

ax[0].hist(clean["pIC50"], bins=50, color="steelblue", edgecolor="white")
ax[0].set_xlabel("pIC50")
ax[0].set_ylabel("Number of molecules")
ax[0].set_title("Distribution of pIC50 (AChE, ChEMBL)")

ax[1].scatter(clean["tanimoto"], clean["pIC50"], s=6, alpha=0.3, color="darkorange")
ax[1].set_xlabel("Tanimoto similarity to donepezil")
ax[1].set_ylabel("pIC50")
ax[1].set_title("Similarity vs potency")

plt.tight_layout()
plt.savefig(f"{WORK}/fig_distributions.png", dpi=150)
plt.show()
```
